Для начала обсудим условия решения задачи:

    * Опыт работы 3 года 2 месяца…

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
import re
def get_work_exp(str_data: str | float) -> int | float:
    if pd.isna(str_data) or str_data == 'Не указано':
            return np.nan
       
    sum_exp = 0
    list_data = str_data.split()[2:]
    pattern = ('год', 'лет', 'месяц')
    for i in range(len(list_data)):
        curr = list_data[i]
        if not(curr.isdigit() or curr.startswith(pattern)):
            break
        elif curr.startswith(('г', 'л')):
            sum_exp += int(list_data[i-1]) * 12
        elif curr.startswith('м'):
            sum_exp += int(list_data[i-1])
    
    return sum_exp
            
                
hh_data['Опыт работы (месяц)'] = hh_data['Опыт работы'].apply(get_work_exp)
hh_data.drop('Опыт работы', axis = 1, inplace=True)
hh_data.head(3)
```

Создадим отдельные признаки **"Город"**, **"Готовность к переезду"**, **"Готовность к командировкам"**. При этом важно учесть:

* Признак **"Готовность к командировкам"** должен иметь два возможных варианта: True или False. Обратите внимание, что возможны несколько вариантов описания готовности к командировкам в признаке "Город, переезд, командировки". Например:
    * … , готов к командировкам , … 
    * … , готова к редким командировкам , …
    * … , не готов к командировкам , …

    Еще один важный факт: при выгрузки данных у некоторых соискателей "потерялась" информация о готовности к командировкам. Давайте по умолчанию будем считать, что такие соискатели не готовы к командировкам.
    
Выполните преобразования и удалите столбец **"Город, переезд, командировки"** из таблицы.

*Совет: обратите внимание на то, что структура текста может меняться в зависимости от указания ближайшего метро. Учите это, если будете использовать порядок слов в своей программе.*


```python
def parse_location_features(str_data: str) -> list[str | bool]:
    million_cities = ['Новосибирск', 'Екатеринбург', 'Нижний Новгород', 'Казань', 'Челябинск', 'Омск', 'Самара', 'Ростов-на-Дону', 'Уфа', 'Красноярск', 'Пермь', 'Воронеж', 'Волгоград' ]
    patterns = ('готов', 'хочу')
    
    list_data = [x.strip() for x in str_data.split(' , ')]
    
    # Удаляем метро если есть
    if list_data[1].startswith('м.'):
        del list_data[1]
    
    result = []
    
    # Город
    if (city := list_data[0]) in ['Москва', 'Санкт-Петербург']: result.append(city)
    elif city in million_cities: result.append('город-миллионник')
    else: result.append('другие')
    
    # Переезд
    result.append(list_data[1].startswith(patterns))
    
    # Командировка
    if len(list_data) > 2:
        result.append(list_data[2].startswith(patterns))
    else:
        result.append(False)
        
    
    return result
    
        
    
```

```python
hh_data[['Город', 'Готовность к переезду', 'Готовность к командировкам']] = hh_data['Город, переезд, командировки'].apply(parse_location_features).to_list()
hh_data.drop('Город, переезд, командировки', axis=1, inplace=True)
hh_data.head(3)
```

```python
def parse_salary(str_data:str)->tuple[int, str]:
    amount, currency = re.search(pattern=r'(\d+\s?\d*)\s(руб|грн|бел\.руб|сум|[A-Z]*)', string=str_data).groups()
    test = {'грн': 'UAH', 'бел.руб':'BYN', 'сум':'UZS'}

    if test.get(currency) is not None: currency = test.get(currency)
        
    return  int(amount), currency
```

```python
# Преобразуем ЗП
hh_data[['ЗП', 'currency']] = hh_data['ЗП'].apply(parse_salary).to_list()
hh_data.head(3)
```

```python
# Объединеям таблицы
hh_exch = hh_data.merge(
    exch_rate,
    left_on=['Обновление резюме', 'currency'],
    right_on=['date', 'currency'],
    how='left'
)
# Заполняем NaN
values_fill = {
    'close': 1,
    'proportion': 1
}
fill_hh_exch = hh_exch.fillna(values_fill)
# считаем новый признак ЗП в рублях
hh_data['ЗП (руб)'] = ((fill_hh_exch['ЗП'] * fill_hh_exch['close']) / fill_hh_exch['proportion'])

# удаляем не нужные столбцы
hh_data.drop(columns=['ЗП', 'currency'], inplace=True)
```

```python
hh_data['Готовность к командировкам'] = hh_data['Готовность к командировкам'].astype(bool)
hh_data['Готовность к переезду'] = hh_data['Готовность к переезду'].astype(bool)

hh_data.info()
```

```python
def outliers_z_score(data, feature, log_scale=False):
    if log_scale:
        x = np.log(data[feature] + 1)
    else:
        x = data[feature]
    mu = x.mean()
    sigma = x.std()
    lower_bound = mu - 3 * sigma
    upper_bound = mu + 3 * sigma
    outliers = data[(x <  lower_bound) | (x > upper_bound)]
    cleaned = data[(x >= lower_bound) & (x <= upper_bound)]
    return outliers, cleaned, (lower_bound, upper_bound)
```
